##Know the Future Today
Having a forecast of the number of claims in the near future provides many benefits including better capital management. Today the department uses several software tools that are traditional PC applications. The chanellege has been sharing the forecasting models to other team members so they can review, improve or validate the models. After a model is approved the model it typically ran on an individual computer where the results then have to be transferred to the main data warehouse. This is an error proned process since if the ETL process on the data warehouse doesn't run or runs late and the forecasting model doesn't use the most recent information and finally getting the forecast results back to the data warehouse often has issues. You are tasked with exploring ways of combining the ML tools directly with
data warehouse to solve all these issues.
