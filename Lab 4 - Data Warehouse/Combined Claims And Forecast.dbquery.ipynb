```sql
select DATE_TRUNC('DAY',DSCHRG_DT) DATE, round(sum(TOT_MDCD_PD_AMT), 0) as PAYMENT_AMOUNT, COUNT(*) as NUM_CLAIMS, TRUE HISTORY
from scott_dev.silver.tmsis_claims tc
where DSCHRG_DT is not null
group by DATE
union all
select ts.discharge_date DATE, round(sum(ts.total_amount)) as PAYMENT_AMOUNT, COUNT(*) NUM_CLAIMS, FALSE HISTORY
from scott_dev.gold.tmsis_forecast ts
group by DATE
```
