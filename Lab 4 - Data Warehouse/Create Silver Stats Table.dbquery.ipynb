```sql
 CREATE TABLE IF NOT EXISTS SILVER.TMSIS_STATS
 AS
 SELECT
    DGNS_CD_1,
    MIN(TOT_MDCD_PD_AMT) AS min_claim_rec_dt,
    MAX(TOT_MDCD_PD_AMT) AS max_claim_rec_dt,
    AVG(TOT_MDCD_PD_AMT) AS avg_claim_rec_dt,
    STDDEV_SAMP(TOT_MDCD_PD_AMT) AS sd_claim_rec_dt
  FROM silver.tmsis_claims
  GROUP BY DGNS_CD_1
```
