## Overview
Genie Agents allow organizations to create designated agents where specific datasets can be provided to users who do not have any SQL or coding expereince or even know what database tables are. They simply ask their questions and Genie provides the answer. While GenAI is used to create SQL queries the answers are the results of the query itself not an answer generated by GenAI so you are not really sure if it is accurate.

Key features typically include:

- Secure access controls for sensitive data
- Answers are returned from SQL queries not GenAI answers
- Ability to share specific datasets without exposing the entire data lake
- Audit capabilities to track who has accessed what data
- Review the SQL queries so you know if correct answers are being returned
- Integration with Databricks' other tools like notebooks and ML capabilities

Genie Agents help solve the challenge of enabling collaboration on data projects while ensuring proper data security and compliance with regulations.


Databricks Genie Agents allows anyone who wants answers from their data to get them without having to have any SQL knowledge.
